```python
# Mengimpor library yang digunakan dalam proses preprocessing
import re
import pandas as pd

# Mengimpor modul Sastrawi untuk menghapus stopword dan mengubah kata berimbuhan menjadi kata dasar
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
# Membuat 5 dokumen bertema penggunaan teknologi dalam pembelajaran mahasiswa
dokumen = [
    "Teknologi sangat membantu mahasiswa dalam proses belajar. Mahasiswa bisa menggunakan laptop untuk mengerjakan tugas kuliah.",
    "Mahasiswa sering menggunakan internet untuk mencari materi perkuliahan. Dengan internet, mencari informasi menjadi lebih mudah dan cepat.",
    "Laptop banyak digunakan mahasiswa untuk membuat tugas kuliah. Selain untuk tugas, laptop juga bisa digunakan untuk mengikuti kelas online.",
    "Aplikasi pembelajaran membantu mahasiswa memahami materi kuliah. Mahasiswa bisa membaca materi dan mengerjakan tugas melalui aplikasi tersebut.",
    "Pembelajaran menggunakan teknologi membuat kegiatan kuliah menjadi lebih praktis. Mahasiswa dapat belajar dari rumah dengan bantuan internet dan perangkat komputer."
]

# Menampilkan jumlah dokumen yang digunakan
print("Jumlah dokumen:", len(dokumen))
```

```python
# Membuat stopword remover Bahasa Indonesia menggunakan Sastrawi
stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()

# Membuat stemmer Bahasa Indonesia menggunakan Sastrawi
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

```python
def preprocess_text(text):
    """
    Melakukan preprocessing teks dengan tahapan:
    1. Case folding
    2. Cleaning
    3. Tokenisasi
    4. Stopword removal
    5. Stemming
    """

    # Case folding: mengubah seluruh huruf menjadi huruf kecil
    text = text.lower()

    # Cleaning: menghapus angka, tanda baca,
    # dan karakter khusus
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Menghapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenisasi: memisahkan teks menjadi beberapa token/kata
    tokens = text.split()

    # Menghapus stopword dari teks
    text_no_stopword = stopword_remover.remove(' '.join(tokens))
    tokens_no_stopword = text_no_stopword.split()

    # Stemming: mengubah kata berimbuhan menjadi kata dasar
    tokens_stemmed = [
        stemmer.stem(token)
        for token in tokens_no_stopword
    ]

    return tokens_stemmed
```

```python
# Menerapkan fungsi preprocessing pada seluruh dokumen
hasil_preprocessing = []

for teks in dokumen:
    hasil = preprocess_text(teks)
    hasil_preprocessing.append(hasil)

# Menampilkan hasil preprocessing setiap dokumen
for i, hasil in enumerate(hasil_preprocessing, start=1):
    print(f"Dokumen {i}:")
    print(hasil)
    print()
```

```python
# Menampilkan perbandingan teks sebelum dan sesudah preprocessing
for i in range(len(dokumen)):
    print(f"===== Dokumen {i + 1} =====")

    # Teks asli sebelum preprocessing
    print("Sebelum:")
    print(dokumen[i])

    # Hasil setelah preprocessing
    print("\nSesudah:")
    print(" ".join(hasil_preprocessing[i]))

    print("\n")
```

```python
# Menghitung jumlah token sebelum dan sesudah preprocessing
data_token = []

for i in range(len(dokumen)):
    # Token sebelum preprocessing
    token_sebelum = re.findall(r'\b[a-zA-Z]+\b', dokumen[i].lower())

    # Token setelah preprocessing
    token_sesudah = hasil_preprocessing[i]

    data_token.append({
        "Dokumen": f"D{i + 1}",
        "Token Sebelum": len(token_sebelum),
        "Token Sesudah": len(token_sesudah)
    })

# Membuat tabel hasil perhitungan token
df_token = pd.DataFrame(data_token)

display(df_token)
```

```python
# Menghitung persentase pengurangan jumlah token
df_token["Pengurangan"] = (
    df_token["Token Sebelum"] - df_token["Token Sesudah"]
)

df_token["Persentase Pengurangan"] = (
    df_token["Pengurangan"] /
    df_token["Token Sebelum"] * 100
)

# Menampilkan hasil dengan pembulatan dua angka desimal
display(
    df_token[
        [
            "Dokumen",
            "Token Sebelum",
            "Token Sesudah",
            "Pengurangan",
            "Persentase Pengurangan"
        ]
    ].round(2)
)
```

```python
# Menampilkan token sebelum dan sesudah preprocessing
for i in range(len(dokumen)):

    # Mengambil token dari teks asli
    token_sebelum = re.findall(
        r'\b[a-zA-Z]+\b',
        dokumen[i].lower()
    )

    print(f"===== Dokumen {i + 1} =====")

    print("Token sebelum:")
    print(token_sebelum)

    print("\nToken sesudah:")
    print(hasil_preprocessing[i])

    print("\n")
```

### Analisis

Preprocessing yang dilakukan membuat teks menjadi lebih bersih dan sederhana. Beberapa bagian seperti huruf kapital, tanda baca, angka, dan kata yang kurang penting dihilangkan agar tidak terlalu banyak mengganggu proses pengolahan data. Stemming juga membuat kata berimbuhan menjadi kata dasar sehingga kata yang mirip lebih mudah dikenali. Dengan begitu, jumlah token berkurang dan data menjadi lebih mudah digunakan untuk proses pencarian dokumen pada sistem Information Retrieval (IR).
